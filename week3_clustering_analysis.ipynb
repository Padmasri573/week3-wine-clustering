# Week 3 — Wine Clustering Analysis

```
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, fcluster
import pandas as pd
```

```
wine=load_wine(as_frame=True)
X=wine.data.copy(); y=wine.target.copy()
print(X.shape)
print('Missing:',X.isna().sum().sum())
print('Duplicates:',X.duplicated().sum())
```

## Standardization

```
X_scaled=StandardScaler().fit_transform(X)
```

## Select k using inertia and silhouette

```
rows=[]
for k in range(2,9):
    km=KMeans(n_clusters=k,random_state=42,n_init=20)
    labels=km.fit_predict(X_scaled)
    rows.append((k,km.inertia_,silhouette_score(X_scaled,labels)))
pd.DataFrame(rows,columns=['k','inertia','silhouette'])
```

## Primary model: k=3

```
km=KMeans(n_clusters=3,random_state=42,n_init=20)
labels=km.fit_predict(X_scaled)
print(silhouette_score(X_scaled,labels))
```

## PCA visualization and hierarchical comparison

```
pca=PCA(n_components=2,random_state=42)
Xp=pca.fit_transform(X_scaled)
Z=linkage(X_scaled,method='ward')
h=fcluster(Z,t=3,criterion='maxclust')-1
print('Hierarchical silhouette:',silhouette_score(X_scaled,h))
```

## Post-hoc validation only

```
print('ARI:',adjusted_rand_score(y,labels))
```
